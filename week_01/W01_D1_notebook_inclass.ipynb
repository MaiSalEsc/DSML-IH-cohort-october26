```python
bill = 23.50

## if bills are bigger than 30 euros usually they get 1 euro discount, if they are btw 20 and 30 get 0.50 and otherwise nothing

if bill >= 30:
    discount = 1
elif bill >=20:
    discount = 0.50
else:
    discount= 0

descounted_bill = bill - discount
descounted_bill
```
