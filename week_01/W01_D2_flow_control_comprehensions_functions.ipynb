## 1 · Indentation: how Python groups code
In Python, a line ending with **`:`** opens a **block**, and the lines of that block are **indented** (4 spaces, or press Tab). The block ends when the indentation stops. Other languages use `{ }`; Python uses the spaces themselves.

```python
if bill > 20:                 # ← the colon opens a block
    print("Big order!")       # ← inside the block (indented)
    print("Add a cookie")     # ← still inside
print("Done")                 # ← outside: always runs
```

> ⚠️ **Classic mistake:** a missing `:` gives a `SyntaxError`; wrong spaces give an `IndentationError`. Colab and VS Code indent for you after a `:`.

## 2 · Decisions: `if / elif / else`
Python checks the conditions **from top to bottom** and runs the **first** block whose condition is `True`. `elif` ("else if") and `else` are optional.

````python
bill = 23.50

if bill >= 30:
    discount = 0.10
elif bill >= 20:
    discount = 0.05
else:
    discount = 0

print(f"Discount: {discount:.0%} → you pay €{bill * (1 - discount):.2f}")
````

🤔 **Predict first:** what discount does a bill of €30 get? And €19.99? Change `bill` above and run it again.

Conditions can be combined (`and`, `or`, `not`) and nested:

````python
item = "latte"
is_student = True

if item in menu:
    price = menu[item]
    if is_student:
        price = price * 0.9
    print(f"{item}: €{price:.2f}")
else:
    print(f"Sorry, we don't sell {item}.")
````

## 3 · `for` loops
A `for` loop **repeats a block once for each element** of a collection. The loop variable (`item` below) takes each value in turn.

```python
for item in collection:
    do something with item
```

````python
for item in ["latte", "tea", "muffin"]:
    print(f"Preparing a {item}...")
print("All done!")
````

### `range()`: repeat a fixed number of times
`range(n)` gives the numbers `0, 1, ..., n-1`. `range(start, stop)` stops **before** `stop` (like slicing).

````python
for i in range(3):
    print("Table", i)

for hour in range(8, 12):
    print(f"{hour}:00 – café open")
````

### The accumulator pattern ⭐
The most useful loop in data work: start from an empty value, then **update it at every step**.

````python
total_items = 0
for item, qty in orders:          # each order is a tuple → unpack it
    total_items += qty
print("Items sold:", total_items)
````

### Looping over a dictionary
`.items()` gives you each key **and** its value:

````python
for item, price in menu.items():
    print(f"{item:<12} €{price:.2f}")
````

🧑‍💻 **Code with me:** compute the **revenue** of the morning: for each order, add `quantity × menu price`. Skip items that are not on the menu (look at `"hot chocolate"`!).

````python
revenue = 0
for item, qty in orders:
    if item in menu:
        revenue += qty * menu[item]
print(f"Revenue: €{revenue:.2f}")   # €46.40
````

> ⚠️ **Classic mistake:** writing `revenue = 0` **inside** the loop resets it at every step. The accumulator starts **before** the loop.

## 4 · List comprehensions
Many loops just **build a new list** from an old one. A **list comprehension** does that in **one readable line**:

```python
new_list = [expression for item in collection]               # transform
new_list = [expression for item in collection if condition]  # transform + filter
```

Read it like English: "*give me `expression` for each `item` in `collection` (if `condition`)*".

````python
prices = list(menu.values())

# With a loop...
prices_with_vat = []
for p in prices:
    prices_with_vat.append(round(p * 1.19, 2))

# ...and with a comprehension: same result, one line
prices_with_vat = [round(p * 1.19, 2) for p in prices]
print(prices_with_vat)
````

Remember the messy order from Notebook 1? Now we can clean **every** item at once:

````python
raw_order = "  Cappuccino, Croissant,MUFFIN  "
items = [x.strip().lower() for x in raw_order.split(",")]
print(items)
````

With a **filter** (`if` at the end), we keep only some elements:

````python
cheap_items = [item for item, price in menu.items() if price < 3]
print(cheap_items)
````

🧑‍💻 **Code with me:** using list comprehensions, (1) make a list of the **quantities** of all orders with 2 or more items, (2) make a list of the **item names** of all orders in UPPERCASE.

> ⚠️ **Classic mistake (style):** comprehensions are for **short** transformations. If you need several lines, nested `if`s or `print`s, a normal loop is clearer. Readability first!

## 5 · Functions
### Why functions? DRY: Don't Repeat Yourself
In section 3 we wrote code to compute the revenue. Tomorrow we'll need it again, and next week for another café. Copy-pasting code is risky: fix a bug in one copy and the others stay wrong. A **function** gives a **name** to a piece of code so you can **reuse** it. You already use many: `print()`, `len()`, `round()`, `sum()`.

### Writing a function
```python
def function_name(argument1, argument2):
    """One line saying what the function does."""
    ...code...
    return result
```
- `def` starts the definition, then the name, the **arguments** in `( )`, and `:`.
- The indented block is the function's body. A `"""docstring"""` explains it.
- `return` sends a result **back** to whoever called the function.

````python
def order_cost(item, qty):
    """Return the cost of qty units of item, using the menu prices."""
    return qty * menu[item]

print(order_cost("latte", 2))
cost = order_cost("muffin", 3)
print(f"3 muffins cost €{cost:.2f}")
````

Defining a function does nothing visible: it only **teaches** Python a new word. The code runs when you **call** it: `order_cost("latte", 2)`.

### `return` vs `print`
`print` only **shows** a value on the screen. `return` **gives it back** so you can store it and keep calculating.

````python
def cost_print(item, qty):
    print(qty * menu[item])

def cost_return(item, qty):
    return qty * menu[item]

a = cost_print("tea", 2)      # shows 5.6 ...
b = cost_return("tea", 2)
print("a =", a, "| b =", b)   # ... but a is None: nothing was returned!
````

### Several arguments and default values
An argument can have a **default value**: if the caller doesn't give it, the default is used. You can also pass arguments **by name** (keyword arguments), which makes calls easier to read.

````python
def apply_discount(price, rate=0.10):
    """Return the price after a discount (10% by default)."""
    return round(price * (1 - rate), 2)

print(apply_discount(20))                # default 10%
print(apply_discount(20, 0.25))          # 25%
print(apply_discount(price=20, rate=0.5))
````

🧑‍💻 **Code with me:** write a function `split_bill(total, people, tip=0.10)` that returns how much each person pays, rounded to cents. Test it with a €47.80 bill for 4 people, with the default tip and with no tip.

````python
def split_bill(total, people, tip=0.10):
    """Return each person's share of the bill, tip included."""
    return round(total * (1 + tip) / people, 2)

print(split_bill(47.80, 4))          # 13.14
print(split_bill(47.80, 4, tip=0))   # 11.95
````

## B2 · Two loop helpers: `enumerate` and `zip`
- `enumerate(list)` gives the **position and the value** → handy to number things.
- `zip(list_a, list_b)` walks through **two lists side by side**.

````python
for position, (item, qty) in enumerate(orders[:3], start=1):
    print(f"Order #{position}: {qty} × {item}")

days = ["Mon", "Tue", "Wed"]
sales = [412.50, 389.90, 455.10]
for day, amount in zip(days, sales):
    print(day, amount)
````

## B3 · `while` loops
A `while` loop repeats **as long as a condition is `True`**. Use it when you **don't know in advance** how many repetitions you need.

````python
coffee_beans_g = 250        # grams of beans left
cups = 0
while coffee_beans_g >= 18:  # one espresso uses 18 g
    coffee_beans_g -= 18
    cups += 1
print(f"{cups} espressos made, {coffee_beans_g} g of beans left")
````

| Use | when |
|---|---|
| `for` | you loop over a known collection or a fixed number of times |
| `while` | you repeat until something happens (stock runs out, target reached, user types "quit") |

> ⚠️ **Classic mistake:** forgetting to update the condition (here `coffee_beans_g -= 18`) creates an **infinite loop**. Stop it with the ⏹ button in Colab (or **Ctrl + C** in a terminal).

````python
savings = 1000
months = 0
while savings <= 1500:
    savings *= 1.03
    months += 1
print(months, round(savings, 2))   # 14 months, €1512.59
````

## B4 · `break` and `continue`
- **`break`** leaves the loop immediately.
- **`continue`** skips the rest of **this** step and goes to the next element.

````python
# Find the first order with 3 or more items, then stop looking
for item, qty in orders:
    if qty >= 3:
        print(f"First big order: {qty} × {item}")
        break
````

````python
# Print only the items we actually sell
for item, qty in orders:
    if item not in menu:
        continue                  # skip unknown items
    print(item, qty)
````

✏️ **Exercise:** go through `orders` and keep a running total of the revenue. Skip items not on the menu with `continue`, and **stop** with `break` as soon as the total goes above €20. Print how many orders it took and the total.

````python
total = 0
count = 0
for item, qty in orders:
    if item not in menu:
        continue
    total += qty * menu[item]
    count += 1
    if total > 20:
        break
print(f"{count} orders → €{total:.2f}")   # 5 orders → €24.80
````

## B5 · Mini-case: the café's daily report ☕📊
Let's combine everything: loops, `if`, dictionaries and f-strings. We want: (1) items sold **per product**, (2) revenue **per product**, (3) the **best seller**, (4) the list of **unknown items**.

````python
sold = {}            # product → quantity
revenue_by_item = {} # product → €
unknown = []

for item, qty in orders:
    if item not in menu:
        unknown.append(item)
        continue
    sold[item] = sold.get(item, 0) + qty                       # counting pattern with .get()
    revenue_by_item[item] = round(revenue_by_item.get(item, 0) + qty * menu[item], 2)

print("Sold:", sold)
print("Revenue:", revenue_by_item)
print("Unknown items:", unknown)
````

`sold.get(item, 0) + qty` means: "the current count for this item, or 0 if it's the first time, plus the new quantity". It's the standard way to **count with a dictionary**.

✏️ **Exercise:** find the **best seller** (the item with the highest quantity) with a loop, and print the report nicely:
```

```

````python
best_item, best_qty = None, 0
for item, qty in sold.items():
    if qty > best_qty:
        best_item, best_qty = item, qty
print(f"Best seller: {best_item} ({best_qty} sold)")
print(f"Total revenue: €{sum(revenue_by_item.values()):.2f}")
# Shortcut you'll see often: max(sold, key=sold.get)
````

## B7 · Scope: what happens inside a function stays inside
Variables created **inside** a function only exist **inside** it (they are *local*).

````python
def add_vat(price):
    vat = 0.19                 # local variable
    return round(price * (1 + vat), 2)

print(add_vat(10))
````

````python
print(vat)                      # NameError: vat only exists inside add_vat
````

## B8 · Handling errors: `try / except` and `raise`
What if someone splits a bill between **0** people, or types `"four"`? The program would crash. With `try / except` we **catch** the error and react to it calmly:

```python
try:
    code that might fail
except SomeError:
    what to do if it fails
```

````python
def safe_split(total, people):
    """Split a bill, with friendly messages for bad inputs."""
    try:
        return round(total / int(people), 2)
    except ZeroDivisionError:
        print("You can't split a bill between 0 people!")
    except ValueError:
        print(f"'{people}' is not a number of people.")

print(safe_split(47.80, 4))
safe_split(47.80, 0)
safe_split(47.80, "four")
````

Sometimes the input is technically valid but makes **no sense** for us, like a negative price. We can **raise** our own error with a clear message:

````python
def checked_price(price):
    if price < 0:
        raise ValueError(f"A price can't be negative (got {price}).")
    return price

print(checked_price(3.40))
````

````python
checked_price(-2)
````

✏️ **Exercise:** write a function `to_price(text)` that converts a text like `"3.40"` into a float. If the conversion fails (e.g. `"three"`), print `"Not a valid price: three"` and return `None`. Test it with both.

````python
def to_price(text):
    """Convert text to a float price, or return None if it isn't a number."""
    try:
        return float(text)
    except ValueError:
        print(f"Not a valid price: {text}")
        return None

print(to_price("3.40"))
print(to_price("three"))
````

## B9 · The mini-case, final version: three small functions
Good code is built from **small functions that each do one thing**. Here is our daily report again, now reusable for any list of orders:

````python
def count_sold(orders, menu):
    """Return {item: quantity} for the items on the menu."""
    sold = {}
    for item, qty in orders:
        if item in menu:
            sold[item] = sold.get(item, 0) + qty
    return sold

def total_revenue(sold, menu):
    """Return the total revenue in euros."""
    return sum(qty * menu[item] for item, qty in sold.items())

def daily_report(orders, menu):
    """Print a short report of the day."""
    sold = count_sold(orders, menu)
    best = max(sold, key=sold.get)
    print(f"Items sold: {sum(sold.values())}")
    print(f"Best seller: {best} ({sold[best]} sold)")
    print(f"Revenue: €{total_revenue(sold, menu):.2f}")

daily_report(orders, menu)
````

Tomorrow, with new orders, we just call `daily_report(tomorrow_orders, menu)`. That's the power of functions.
