| `+ - *` | `/` | `//` | `%` | `**` |
|---|---|---|---|---|
| usual | always float | whole division | remainder | power |

````python
print(7 / 2, 7 // 2, 7 % 2, 2 ** 10)

price_cappuccino = 3.40
price_croissant = 2.20
bill = 2 * price_cappuccino + price_croissant
print(bill, type(bill))

revenue = 0
revenue += bill          # same as revenue = revenue + bill
print(revenue)
````

## 2 · Strings and f-strings
Text between quotes. **Methods** are called with a dot: `.strip()`, `.lower()`, `.upper()`, `.title()`, `.replace(a, b)`, `.split(sep)`.

````python
cafe = "Café Ironhack"
print(len(cafe), cafe.upper())

raw_order = "  Cappuccino, Croissant,MUFFIN  "
print(raw_order.strip().lower().split(","))   # spaces left inside → loops fix it later
````

## 6 · Decisions: `if / elif / else`
A line ending in **`:`** opens a block; the block is **indented** (4 spaces). Python runs the **first** true branch.

````python
bill = 23.50
if bill >= 30:
    discount = 0.10
elif bill >= 20:
    discount = 0.05
else:
    discount = 0
print(f"Discount {discount:.0%} → you pay €{bill * (1 - discount):.2f}")
````

🤔 **Predict:** what about €30? €19.99? Change `bill` and re-run.

🧑‍💻 **Code with me:** size `S` = menu price, `M` = +€0.50, `L` = +€1.00, else `"Unknown size"`. Large cappuccino?

````python
size = "L"
price = menu["cappuccino"]
if size == "S":
    print(price)
elif size == "M":
    print(price + 0.50)
elif size == "L":
    print(price + 1.00)
else:
    print("Unknown size")
````

## 7 · `for` loops
Repeat a block **once per element**. `range(n)` gives `0 … n-1`.

````python
for item in ["latte", "tea", "muffin"]:
    print(f"Preparing a {item}...")

for hour in range(8, 11):
    print(f"{hour}:00 – open")

for item, price in list(menu.items())[:3]:   # dict: key and value
    print(f"{item:<12} €{price:.2f}")
````

### ⭐ The accumulator pattern
Start **before** the loop, update **inside** it. Here are this morning's orders (a list of tuples):

````python
orders = [("latte", 2), ("croissant", 1), ("espresso", 1), ("latte", 1), ("muffin", 3),
          ("cappuccino", 2), ("tea", 1), ("croissant", 2), ("latte", 2), ("hot chocolate", 1)]

total_items = 0
for item, qty in orders:
    total_items += qty
print("Items ordered:", total_items)
````

🧑‍💻 **Code with me:** compute the morning's **revenue**, skipping items not on the menu.

````python
revenue = 0
for item, qty in orders:
    if item in menu:
        revenue += qty * menu[item]
print(f"Revenue: €{revenue:.2f}")
````

**Bonus — counting with a dictionary** (the most useful pattern in data work):

````python
sold = {}
for item, qty in orders:
    sold[item] = sold.get(item, 0) + qty
print(sold)
````

> ⚠️ `revenue = 0` **inside** the loop resets it every step.

## 8 · List comprehensions
A loop that builds a list, in one line: `[expression for x in collection if condition]`

````python
prices = list(menu.values())
with_vat = [round(p * 1.19, 2) for p in prices]                 # transform
cheap = [item for item, price in menu.items() if price < 3]      # filter
clean = [x.strip().lower() for x in raw_order.split(",")]        # the messy order, fixed!
print(with_vat, cheap, clean, sep="\n")
````

🧑‍💻 **Code with me:** the quantities of orders with 2+ items.

````python
[qty for item, qty in orders if qty >= 2]
````

## 9 · Functions
Give a **name** to code so you can reuse it (DRY: Don't Repeat Yourself).
```python
def name(arg1, arg2=default):
    """What it does."""
    return result
```

````python
def order_cost(item, qty):
    """Return the cost of qty units of item."""
    return qty * menu[item]

print(order_cost("latte", 2))
````

`print` only **shows**; `return` **gives back** a value you can keep using:

````python
def cost_print(item, qty):
    print(qty * menu[item])

a = cost_print("tea", 2)
print("a =", a)        # None: nothing was returned!
````

🧑‍💻 **Code with me:** `split_bill(total, people, tip=0.10)` → each person's share.

````python
def split_bill(total, people, tip=0.10):
    """Each person's share, tip included, rounded to cents."""
    return round(total * (1 + tip) / people, 2)

print(split_bill(47.80, 4))          # default tip
print(split_bill(47.80, 4, tip=0))   # no tip
````

**Putting it all together:** a reusable daily report.

````python
def daily_revenue(orders, menu):
    """Total revenue for the items on the menu."""
    return sum(qty * menu[item] for item, qty in orders if item in menu)

afternoon = [("tea", 3), ("muffin", 2), ("latte", 1)]
print(f"Morning: €{daily_revenue(orders, menu):.2f} | Afternoon: €{daily_revenue(afternoon, menu):.2f}")
````
