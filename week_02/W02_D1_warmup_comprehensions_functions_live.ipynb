## 1 · List comprehensions

### From a loop to one line
Very often a loop just **builds a new list** from an old one. Here we convert steps into thousands of steps:

````python
# The loop way: 3 lines
````

A **list comprehension** does exactly the same in **one line**:

```python
new_list = [expression for item in collection]
```

Read it from the inside out, like English: *"give me `s / 1000` **for each** `s` **in** `steps`"*.

### Filter: keep only some elements
Add an `if` **at the end** to keep only the items that pass a test:

```python
new_list = [expression for item in collection if condition]
```

Which days did the person walk **at least 8,000 steps**?

### Choose between two values: `if ... else` **before** the `for`
To **label** every element (instead of throwing some away), put `value_if_true if condition else value_if_false` at the **start**:

> 💡 **Where does the `if` go?**
> - `[x for x in data if x > 0]` → **filter**: the new list can be **shorter**.
> - `[a if x > 0 else b for x in data]` → **choose**: the new list has the **same length**.

## 🧭 Recap

| Pattern | Example |
|---|---|
| transform | `[s / 1000 for s in steps]` |
| filter | `[s for s in steps if s >= 8000]` |
| choose | `["active" if s >= 8000 else "rest" for s in steps]` |
| function + default | `def steps_to_km(n_steps, stride_m=0.75):` |
| function inside a comprehension | `[steps_to_km(s) for s in steps]` |
| return several values | `return average, good_days, percent` |
